# Evaluate a lifelines Cox model

Fit a Cox proportional hazards model to GBSG2, inspect its individual survival curves,
and evaluate discrimination, prediction error, and calibration on a held-out split.
The adapter accepts lifelines' DataFrame directly: **rows are times, columns are subjects**.

Run the cells from top to bottom in a Python environment with SurvivalEVAL installed
(`python -m pip install -e .` from the repository root). The next cell installs this
example's optional model dependency into the active notebook kernel.

Throughout, **1 / True means an observed event; 0 / False means right censoring**.
An observed time is either an event time or the last known event-free time.
Training outcomes are supplied separately to estimate censoring weights and other
training-distribution quantities; held-out outcomes are used for evaluation only.

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from lifelines import CoxPHFitter
from lifelines.datasets import load_gbsg2
from sklearn.model_selection import train_test_split

from SurvivalEVAL import LifelinesEvaluator

plt.rcParams.update({"figure.figsize": (8, 4), "figure.dpi": 110,
                     "axes.spines.top": False, "axes.spines.right": False,
                     "axes.grid": True, "grid.alpha": 0.2})
```

## Load and split GBSG2

The dataset has 686 subjects, eight predictors, and recurrence-free follow-up in days.
Despite its name, **`cens=1` indicates an event** in this dataset. Binary predictors
are coded as 0/1, and tumour grade is treated as an ordinal predictor (I, II, III).
A stratified 70/30 split preserves the event fraction, with a fixed random seed.

```python
gbsg2 = load_gbsg2()
gbsg2["horTh"] = gbsg2["horTh"].map({"no": 0, "yes": 1})
gbsg2["menostat"] = gbsg2["menostat"].map({"Pre": 0, "Post": 1})
gbsg2["tgrade"] = gbsg2["tgrade"].map({"I": 1, "II": 2, "III": 3})
train, test = train_test_split(gbsg2, test_size=0.3, stratify=gbsg2["cens"], random_state=42)
train_times, train_events = train["time"].to_numpy(), train["cens"].to_numpy()
test_times, test_events = test["time"].to_numpy(), test["cens"].to_numpy()
pd.DataFrame({"Subjects": [len(train), len(test)],
              "Observed event fraction": [train_events.mean(), test_events.mean()]},
             index=["Train", "Test"])
```

## Fit on training subjects; predict for test subjects

The test outcomes do not participate in fitting. Cox models share a baseline hazard;
differences between individual curves arise through their covariates. This demonstration
fits a model without tuning its predictors or checking the proportional hazards assumption.

```python
model = CoxPHFitter().fit(train, duration_col="time", event_col="cens")
survival = model.predict_survival_function(test.drop(columns=["time", "cens"]))
assert survival.columns.equals(test.index)
evaluator = LifelinesEvaluator(survival, test_times, test_events, train_times, train_events)
print(f"Prediction DataFrame: {survival.shape[0]} time points × {survival.shape[1]} test subjects")
survival.iloc[:5, :5]
```

## Inspect the predictions before scoring

Each line below describes one held-out subject, rather than the survival curve of
a group. Survival probabilities must stay in [0, 1] and cannot increase with time.
The vertical line marks our fixed evaluation horizon. Predictions and test outcomes
must remain in exactly the same subject order.

```python
grid = evaluator.time_coordinates
curves = evaluator.pred_survs
assert curves.shape == (len(test_times), len(grid))
assert np.isfinite(curves).all()
assert ((curves >= 0) & (curves <= 1)).all()
assert (np.diff(curves, axis=1) <= 1e-8).all()
horizon = 1000.0
integration_grid = np.linspace(0, horizon, 80)

fig, ax = plt.subplots()
for position in range(5):
    ax.plot(grid, curves[position], label=f"Test subject {position + 1}")
ax.axvline(horizon, color="0.35", linestyle=":", label="Evaluation horizon")
ax.set(xlabel="Time (days)", ylabel="Predicted survival S(t)",
       title="Individual survival predictions on held-out data", ylim=(-0.02, 1.02))
ax.legend(ncol=2, fontsize=9)
fig.tight_layout()
plt.show()
```

## Discrimination: can the predictions order subjects?

Harrell's concordance compares subjects with observable event ordering. The evaluator
uses predicted median survival times: a **larger predicted time means longer survival**.
`ties="Risk"` gives half credit to tied predictions. A C-index near 0.5 corresponds to
chance ordering; higher values are better. It does not measure probability accuracy.

The time-dependent AUC asks whether risk at 1000.0 days ranks cases above controls,
using IPCW to account for censoring.

```python
c_index, concordant, comparable = evaluator.concordance(method="Harrell", ties="Risk")
auc = evaluator.auc(target_time=horizon, method="IPCW")
print(f"Harrell C-index: {c_index:.3f} ({concordant:g} / {comparable:g} comparable pairs)")
print(f"IPCW AUC at {horizon:g}: {auc:.3f}")
```

## Prediction error: probabilities and event times

The IPCW Brier score measures squared probability error at a horizon. IBS averages
this error over **[0, 1000.0] days**, using the explicit grid below. Lower is better;
IBS values from different integration intervals are not directly comparable. We choose
a horizon within the training follow-up rather than extrapolating censoring weights
into the tail. Marginal IPCW assumes censoring is independent of the event time.

MAE-Hinge penalizes an event's absolute error and, for a censored subject, only a
prediction earlier than the censoring time. MAE-Pseudo_obs uses training-distribution
information to account for the unknown event times. These are different estimands:
censoring times are not treated as observed event times. Median predictions can also
depend on tail extrapolation when a predicted curve does not cross 0.5 on its grid.

```python
brier = evaluator.brier_score(target_time=horizon)
brier_curve = evaluator.brier_score_multiple_points(integration_grid)
ibs = evaluator.integrated_brier_score(target_times=integration_grid)
mae_hinge = evaluator.mae(method="Hinge")
mae_po = evaluator.mae(method="Pseudo_obs")
log_mse = evaluator.mse(method="Hinge", log_scale=True)
assert np.isfinite([c_index, auc, brier, ibs, mae_hinge, mae_po, log_mse]).all()

fig, ax = plt.subplots()
ax.plot(integration_grid, brier_curve, color="tab:blue", linewidth=2)
ax.fill_between(integration_grid, brier_curve, alpha=0.15, color="tab:blue")
ax.set(xlabel="Time (days)", ylabel="IPCW Brier score (lower is better)",
       title=f"Prediction error over the chosen interval: IBS = {ibs:.3f}")
fig.tight_layout()
plt.show()
pd.DataFrame({
    "Value": [c_index, auc, brier, ibs, mae_hinge, mae_po, log_mse],
    "Interpretation": ["Higher is better", "Higher is better", "Lower is better",
                       "Lower is better", "days; lower is better",
                       "days; lower is better", "Squared log-time error; lower is better"],
}, index=["Harrell C-index", f"AUC at {horizon:g}", f"Brier at {horizon:g}",
          f"IBS [0, {horizon:g}]", "MAE-Hinge", "MAE-Pseudo_obs", "MSE-Hinge, log scale"])
```

## Calibration: do probabilities agree with outcomes?

**One-calibration** compares predicted event risk `1 - S(horizon)` with a Kaplan–Meier
event estimate in each risk group. Five groups keep this demonstration relatively coarse.
Points above the diagonal indicate underestimated event risk; points below indicate
overestimated risk. The test uses Greenwood variances; a group with zero variance can
make its p-value undefined (`NaN`).

**D-calibration** assesses the distribution of survival probabilities at event times,
allocating censored observations across their possible probability bins. It concerns the
whole predicted distribution rather than a single time horizon.

A small p-value is evidence against the corresponding calibration hypothesis. A large
p-value means the test did not detect a departure, **not proof of calibration**. Inspect
the plots as well: p-values depend on sample size and grouping.

```python
one_p, one_details = evaluator.one_calibration(
    target_time=horizon, num_bins=5, return_details=True
)
fig, ax = one_details["histogram_plot"]
ax.set_title(f"Grouped event risk at {horizon:g}; one-calibration p = {one_p:.3g}")
fig.tight_layout()
fig, ax = one_details["pp_plot"]
ax.set(xlabel="Mean predicted event risk", ylabel="Kaplan–Meier event estimate",
       title=f"Fixed-time calibration at {horizon:g}")
fig.tight_layout()
plt.show()

d_p, d_details = evaluator.d_calibration(num_bins=10, return_details=True)
fig, ax = d_details["histogram_plot"]
ax.set_title(f"D-calibration across the survival distribution; p = {d_p:.3g}")
fig.tight_layout()
plt.show()
print(f"One-calibration p-value: {one_p:.4g}; D-calibration p-value: {d_p:.4g}")
```

These scores describe this fitted model on one held-out split. Use the same split,
evaluation horizon, censoring convention, and interpolation method when comparing
models; use repeated resampling or an external test set to assess uncertainty.
